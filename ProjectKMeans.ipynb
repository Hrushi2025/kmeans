```python
from sklearn.cluster import KMeans
```

```python
kmeans= KMeans(n_clusters=2)
```

```python
kmeans.fit(df.drop('private',axis=1))
```

```python
kmeans.cluster_centers_
```

```python
# evaluations
# create a new column for df as 'cluster' which is a 1 for private school and 0 for public school
def converter(private):
    if private == 'Yes':
        return 1
    else:
        return 0
```

```python
df['cluster']=df['private'].apply(converter)
```

```python
df.head() # now we have added a column cluster where we can 1 and 0 
```

```python
#now we are going to do confusion metrix to check how well kmeans work without giving any lables
from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(confusion_matrix(df['cluster'],kmeans.labels_))
print('\n')
print(classification_report(df['cluster'],kmeans.labels_))
```
