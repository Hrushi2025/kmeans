```python
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_blobs
```

```python
data =make_blobs(n_samples=200,
    n_features=2,
    centers=4,
    cluster_std=1.8,
    random_state=101)
```

```python
data[0].shape # its 200 by 2 data 
```

```python
plt.scatter(data[0][:,0],data[0][:,1],c=data[1],cmap='rainbow') #first index of first row and first index of 2nd column 
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=4) # here we have assigned how many cluster you want make try there 3,4,6 anything 
```

```python
kmeans.fit(data[0])
```

```python
kmeans.cluster_centers_
```

```python
kmeans.labels_ #labels are compalsury for kmeans it is not there your done with this 
```

```python
fig,(ax1,ax2)= plt.subplots(1,2,sharey=True,figsize=(10,6)) #two graphs are created here like subplots with two axis we are grabbing here 
ax1.set_title('K Means')
ax1.scatter(data[0][:,0],data[0][:,1],c=kmeans.labels_,cmap='rainbow') #it is also orignal data but used kemans 
ax2.set_title('Orignal Data')
ax2.scatter(data[0][:,0],data[0][:,1],c=data[1],cmap='rainbow')#it is actual orignal data with its own color 
```
